```python
# Q1. Find the mean of 12, 15, 18, 25
numbers = [12, 15, 18, 25]

mean = sum(numbers) / len(numbers)

print("\nQ1. Mean")
print("Numbers:", numbers)
print("Mean =", mean)
```

```python
# Q2. Find the median of 3, 5, 8, 10, 14
numbers = [3, 5, 8, 10, 14]
numbers.sort()

n = len(numbers)
median = numbers[n // 2] if n % 2 != 0 else (
    numbers[n // 2 - 1] + numbers[n // 2]
) / 2

print("\nQ2. Median")
print("Numbers:", numbers)
print("Median =", median)
```

```python
# Q3. Find the mode of 2, 4, 4, 5, 7, 7, 7
from collections import Counter

numbers = [2, 4, 4, 5, 7, 7, 7]
frequency = Counter(numbers)
highest_frequency = max(frequency.values())

modes = [
    number for number, count in frequency.items()
    if count == highest_frequency
]

print("\nQ3. Mode")
print("Numbers:", numbers)
print("Mode =", modes)
print("Frequency =", highest_frequency)
```

```python
# Q4. Find the range of 12, 19, 24, 31
numbers = [12, 19, 24, 31]

data_range = max(numbers) - min(numbers)

print("\nQ4. Range")
print("Numbers:", numbers)
print("Range =", data_range)
```

```python
# Q5. If Q1 = 8 and Q3 = 20, calculate IQR
Q1 = 8
Q3 = 20

IQR = Q3 - Q1

print("\nQ5. Interquartile Range (IQR)")
print("IQR =", IQR)
```

```python
# Q6. If Q1 = 8 and IQR = 12,
# calculate the upper outlier fence
Q1 = 8
IQR = 12

Q3 = Q1 + IQR
upper_fence = Q3 + 1.5 * IQR

print("\nQ6. Upper Outlier Fence")
print("Q3 =", Q3)
print("Upper Fence =", upper_fence)
```

```python
# Q7. Mean = 60, SD = 5, score = 70
mean = 60
sd = 5
score = 70

z_score = (score - mean) / sd

print("\nQ7. Z-Score")
print("Z-score =", z_score)
```

```python
# Q8. Interpret correlation of -0.9
correlation = -0.9

print("\nQ8. Correlation")
print("Correlation =", correlation)
print("Interpretation: Very strong negative linear correlation.")
```

```python
# Q9. Fair die: probability of number greater than 4
possible_outcomes = [1, 2, 3, 4, 5, 6]
favourable_outcomes = [
    number for number in possible_outcomes if number > 4
]

probability = len(favourable_outcomes) / len(possible_outcomes)

print("\nQ9. Probability: Die")
print("Favourable outcomes:", favourable_outcomes)
print("Probability =", probability)
```

```python
# Q10. Toss a fair coin twice: probability of two Heads
outcomes = ["HH", "HT", "TH", "TT"]
two_heads = [outcome for outcome in outcomes if outcome == "HH"]

probability = len(two_heads) / len(outcomes)

print("\nQ10. Probability: Two Heads")
print("Possible outcomes:", outcomes)
print("Probability =", probability)
```

```python
# Q11. Standard deck: probability of drawing an Ace
total_cards = 52
aces = 4

probability = aces / total_cards

print("\nQ11. Probability: Ace")
print("Probability =", probability)
print("Percentage =", probability * 100, "%")
```

```python
# Q12. If P(A) = 0.25, find P(not A)
P_A = 0.25

P_not_A = 1 - P_A

print("\nQ12. Complement Probability")
print("P(not A) =", P_not_A)
```

```python
# Q13. Standardization vs Min-Max Normalization
print("\nQ13. Standardization vs Normalization")
print("Standardization: (x - mean) / standard deviation")
print("Usually gives data mean 0 and standard deviation 1.")
print("Min-max normalization: (x - minimum) / (maximum - minimum)")
print("Usually scales data between 0 and 1.")
```
